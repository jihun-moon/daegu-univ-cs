```
import seaborn as sns

iris_dataset = sns.load_dataset('iris')
iris_dataset.head()
```

**데이터 설명:**

`iris_dataset`의 열들은 붓꽃의 특정 부분에 대한 측정값을 나타냅니다.

*   `sepal_length`: 꽃받침의 길이(센티미터)
*   `sepal_width`: 꽃받침의 너비(센티미터)
*   `petal_length`: 꽃잎의 길이(센티미터)
*   `petal_width`: 꽃잎의 너비(센티미터)

이 측정값들은 서로 다른 붓꽃 종을 구별하는 데 사용됩니다.

```
sns.countplot(x='species', data=iris_dataset)
```

```
sns.scatterplot(x='petal_length', y='petal_width', hue='species', data=iris_dataset)
```

```
sns.pairplot(iris_dataset, hue='species', palette='coolwarm')
```

```
sns.swarmplot(x='species', y='petal_width', data=iris_dataset)
```

```
setosa = iris_dataset[iris_dataset['species'] == 'setosa']
versicolor = iris_dataset[iris_dataset['species'] == 'versicolor']
virginica = iris_dataset[iris_dataset['species'] == 'virginica']
sns.distplot(setosa['petal_length'], hist=False, label='setosa')
sns.distplot(versicolor['petal_length'], hist=False, label='versicolor')
sns.distplot(virginica['petal_length'], hist=False, label='virginica')
```
