```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10)
y = np.cos(x)
z = np.sin(x)

plt.plot(x, y, x, z)
plt.xlabel('Time Axis')
plt.ylabel('Magnitude')
plt.title('Sine & Cosine Waves')
plt.legend(['y=cos(x)', 'y=sin(x)'], loc=0)
plt.show()
```
