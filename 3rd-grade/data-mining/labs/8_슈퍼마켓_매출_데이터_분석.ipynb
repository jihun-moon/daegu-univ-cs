```
import pandas as pd
df = pd.read_csv('./drive/MyDrive/supermarket_sales.csv')
df.head()
```

```
df['Total'].plot.hist(bins=30)
```

```
b = df.groupby('Branch')['Total'].sum()
b.plot.bar(color=['blue', 'green', 'red'])
```

```
df['Time'] = pd.to_datetime(df['Time'])
df['Hour'] = df['Time'].dt.hour
h = df.groupby('Hour')['Total'].sum()
h.plot.line(marker = 'o')
```

```
p = df.groupby('Payment')['Total'].sum()
p.plot.pie(autopct='%1.1f%%', ylabel=' ')
```
