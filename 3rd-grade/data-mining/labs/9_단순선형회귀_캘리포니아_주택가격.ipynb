```
# 필요한 라이브러리를 가져옵니다.
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# California Housing 데이터셋을 로드합니다.
housing = fetch_california_housing()
# 데이터와 타겟의 형태를 출력합니다.
print(housing.data.shape, housing.target.shape)
# 특징(feature) 이름을 출력합니다.
print(housing.feature_names)
```

```
# 데이터를 학습 세트와 테스트 세트로 80:20 비율로 분할합니다.
# 여기서는 'MedInc' (Median Income) 특징만 사용합니다 (첫 번째 열).
X_train, X_test, y_train, y_test = train_test_split(housing.data[:, 0:1], housing.target, test_size=0.2, random_state=42)
# 분할된 데이터 세트의 형태를 출력합니다.
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

```
# LinearRegression 모델을 초기화합니다.
regr = LinearRegression()
# 학습 데이터를 사용하여 모델을 학습시킵니다.
regr.fit(X_train, y_train)
```

```
# 학습 데이터에 대한 모델의 R-제곱 점수를 계산하여 출력합니다.
regr.score(X_train, y_train)
```

```
# 학습된 선형 회귀 모델의 계수 (기울기)를 출력합니다.
print(regr.coef_)
# 학습된 선형 회귀 모델의 절편을 출력합니다.
print(regr.intercept_)
```

```
# LinearRegression 모델을 다시 초기화하고 학습시킵니다.
regr = LinearRegression()
regr.fit(X_train, y_train)
# 학습 데이터에 대한 R-제곱 점수를 출력합니다.
print(regr.score(X_train, y_train))

# 테스트 데이터를 사용하여 예측을 수행합니다.
y_pred = regr.predict(X_test)
# 테스트 데이터에 대한 Mean Squared Error (평균 제곱 오차)를 출력합니다.
print(mean_squared_error(y_test, y_pred))
```
