```python
x = np.arange(-5, 6, 0.5)

y1 = 3 * x**2 - 2
y2 = 4 * x + 15
y3 = 2 * x**3 + x
y4 = np.cos(x)

plt.subplot(2, 2, 1)
plt.plot(x, y1)
plt.title('y = 3x^2 - 2')

plt.subplot(2, 2, 2)
plt.plot(x, y2)
plt.title('y = 4x + 15')

plt.subplot(2, 2, 3)
plt.plot(x, y3)
plt.title('y = 2x^3 + x')

plt.subplot(2, 2, 4)
plt.plot(x, y4)
plt.title('y = cos(x)')

plt.tight_layout()
plt.show()
```

```python
from sklearn.datasets import fetch_openml
mnist = fetch_openml('mnist_784', version=1)
mnist.keys()
```

```python
print(mnist.data.shape)
print(mnist.target.shape)
```

```python
import matplotlib.pyplot as plt
x = mnist.data
y = mnist.target
digit = x.iloc[3000]
digit_image = digit.values.reshape(28, 28)
plt.imshow(digit_image, cmap='binary')
```

```python
def plot_mnist_image(images, labels, num_rows=2, num_cols=5):
  fig, axes = plt.subplots(num_rows, num_cols, figsize=(10, 5))
  axes = axes.ravel()
  for i in range(num_rows * num_cols):
    axes[i].imshow(images.iloc[i].values.reshape(28, 28), cmap='gray')
    axes[i].set_title(f'Label: {labels.iloc[i]}')
    axes[i].axis('off')
  plt.subplots_adjust(hspace=0.5, wspace=0.5)
  plt.show()

plot_mnist_image(x, y, num_rows=10, num_cols=10)
```
