```
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
data = pd.read_csv('./drive/MyDrive/MELBOURNE_HOUSE_PRICES_LESS.csv')
data.head()
```

```
data.info()
```

```
cols = ['Suburb', 'Rooms', 'Type', 'Method', 'SellerG', 'Regionname', 'Propertycount', 'Distance',
        'CouncilArea', 'Price']
data = data[cols]
data[cols] = data[cols].fillna({'Propertycount':0, 'Distance':0})
data.dropna(inplace=True)
data = pd.get_dummies(data, drop_first=True)
feature = data.drop('Price', axis=1)
target = data['Price']
feature_train, feature_test, target_train, target_test = train_test_split(feature, target, test_size=0.2, random_state=42)
```

```
# 1. Linear Regression
reg = LinearRegression().fit(feature_train, target_train)
print('Linear Train= ', reg.score(feature_train, target_train))
print('Linear Test = ', reg.score(feature_test, target_test))

# 2. Lasso (reg -> lasso 로 변경해야 함)
lasso = Lasso(alpha=50, max_iter=100, tol=0.1).fit(feature_train, target_train)
print('Lasso Train = ', lasso.score(feature_train, target_train)) # 여기!
print('Lasso Test  = ', lasso.score(feature_test, target_test)) # 여기!

# 3. Ridge (reg -> ridge 로 변경해야 함)
ridge = Ridge(alpha=50, max_iter=100, tol=0.1).fit(feature_train, target_train)
print('Ridge Train = ', ridge.score(feature_train, target_train)) # 여기!
print('Ridge Test  = ', ridge.score(feature_test, target_test)) # 여기!

# 4. ElasticNet (reg -> elastic 으로 변경해야 함)
elastic = ElasticNet(random_state=42).fit(feature_train, target_train)
print('Elastic Train= ', elastic.score(feature_train, target_train)) # 여기!
print('Elastic Test = ', elastic.score(feature_test, target_test)) # 여기!
```
